1. Postoje li razlike između shizofrenih i neshizofrenih počinitelja kaznenih djela ubojstva i ubojstva u pokušaju s obzirom na sociodemografska obilježja? – dob, spol, obrazovanje, bračni status;

```python
import pandas as pd
from scipy import stats

# Pretpostavljamo da je "kat" stupac koji označava shizofrene vs neshizofrene (npr. 'NESHIZOFRENI' ili 'SHIZOFRENI')
# Filtriramo podatke u dvije grupe
n_shizofreni = df[df['kat'] == 'NESHIZOFRENI']
shizofreni = df[df['kat'] == 'SHIZOFRENI']
```

```python
# prompt: give me median for 'dob' per each category ('kat')

# Calculate the median 'dob' for each category ('kat')
median_dob_by_kat = df.groupby('kat')['dob'].median()

median_dob_by_kat
```

```python
# prompt: test if distributions are normal for variable  'dob' by 'kat'

from scipy.stats import shapiro

# Test normality for 'dob' in 'NESHIZOFRENI' group
stat_n, p_n = shapiro(n_shizofreni['dob'].dropna())
print(f"Shapiro-Wilk test for NESHIZOFRENI (dob): Statistic={stat_n:.3f}, p-value={p_n:.3f}")

# Test normality for 'dob' in 'SHIZOFRENI' group
stat_s, p_s = shapiro(shizofreni['dob'].dropna())
print(f"Shapiro-Wilk test for SHIZOFRENI (dob): Statistic={stat_s:.3f}, p-value={p_s:.3f}")

# Interpretation
alpha = 0.05
if p_n > alpha:
  print('Sample of NESHIZOFRENI looks Gaussian (fail to reject H0)')
else:
  print('Sample of NESHIZOFRENI does not look Gaussian (reject H0)')

if p_s > alpha:
  print('Sample of SHIZOFRENI looks Gaussian (fail to reject H0)')
else:
  print('Sample of SHIZOFRENI does not look Gaussian (reject H0)')
```

```python
u_stat, p_value = stats.mannwhitneyu(n_shizofreni['dob'], shizofreni['dob'], alternative='two-sided')

print(f"Mann-Whitney U test statistika: {u_stat}, P-vrijednost: {p_value}")
```

##Problem 2
2. Postoje li razlike između shizofrenih i neshizofrenih počinitelja kaznenih djela ubojstva i ubojstva u pokušaju s obzirom na kriminološka obilježja? – način počinjenja/sredstvo, odnos sa žrtvom, konstalativni faktori (odnose sa drogom/alkoholom, konfliktne sit i sukob/svadu), agresivnost, impulzivnost, vrijeme počinjenja djela, mjesto počinjenja, supočiniteljstvo, prisutnost alkohola/droge;

```python
import pandas as pd
from scipy.stats import chi2_contingency

# Pretpostavka: df je vaš DataFrame s podacima, a 'kat' je kategorija prema kojoj se testira

# Definirajte varijable za analizu
variables = ["sredstvo", "odnos", "ovisnost", "sukob", "agres", "impul", "vrijeme", "mjesto", "supoci", "alkdrog"]

# Kreirajte DataFrame za rezultate hi-kvadrat testa
chi2_results = pd.DataFrame(columns=['Variable', 'χ²', 'p-value', 'df'])

# Iterirajte kroz varijable i provedite analizu
for var in variables:
    # Filtriraj podatke ovisno o uvjetima za svaku varijablu:
    if var == "sukob":
        # Za varijablu 'ovisnost', izbacujemo podatke s vrijednošću '2'
        filtered_df = df[df[var] != '2']
    elif var in ["impul", "agres", "alkdrog"]:
        # Za varijable 'sukob' i 'agres', izbacujemo podatke s vrijednošću 'nema podataka'
        filtered_df = df[df[var] != 'nema podataka']
    else:
        # Za ostale varijable, ne filtriramo podatke
        filtered_df = df

    try:
        # Kreiranje kontingencijske tablice
        contingency_table = pd.crosstab(filtered_df['kat'], filtered_df[var])

        # Provedite hi-kvadrat test
        chi2, p, dof, expected = chi2_contingency(contingency_table)

        # Dodajte rezultate u DataFrame
        chi2_results = pd.concat([
            chi2_results,
            pd.DataFrame({'Variable': [var], 'χ²': [chi2], 'p-value': [p], 'df': [dof]})
        ], ignore_index=True)
    except ValueError:
        print(f"Variable '{var}' has insufficient data for Chi-square analysis and will be skipped.")

# Zaokružite vrijednosti za preglednost
chi2_results = chi2_results.round(2)

# Ispišite rezultate
print("Chi-square test results:")
chi2_results
```

```python
# Opcionalno: Prikaz kontingencijske tablice za svaku varijablu
var = 'alkdrog'
if var == "sukob":
    filtered_df = df[df[var] != '2']
elif var in ["impul", "agres", "alkdrog"]:
    filtered_df = df[df[var] != 'nema podataka']
else:
    filtered_df = df


contingency_table = pd.crosstab(filtered_df['kat'], filtered_df[var])
print(f"\nContingency table for {var}:")
contingency_table
```
